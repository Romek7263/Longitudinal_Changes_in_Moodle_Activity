```python
def simple_analyze(column):
    folders = ["2019", "2021", "2022", "2023", "2024", "2025", "2026"]
    for folder in folders:
        data = pd.read_csv('tcs_dataset/data/' + folder + '/grades.csv')
        #print(folder)
        #print(len(data))
        print(folder + " " + column + " " + str(data[column].mean()))
simple_analyze("homework_02_est")
simple_analyze("final_exam_attempt_1_eng")
simple_analyze("final_exam_attempt_1_est")
```

```python


folders = ["2019", "2021", "2022", "2023", "2024", "2025", "2026"]
count = 0
acc = 0
for folder in folders:
    data = pd.read_csv('tcs_dataset/data/' + folder + '/grades.csv')
    print(folder)
    print(len(data))
    print("hw1_eng " + str(data["homework_01_eng"].mean()))
    print("hw2_eng " + str(data["homework_02_eng"].mean()))
    acc += data["homework_02_eng"].mean() * len(data)
    count += len(data)
    print("hw3_eng " + str(data["homework_03_eng"].mean()))
    print(data["final_exam_attempt_1_eng"].mean())
    print("-----------")

mean = acc / count
print("hw2_eng average " + str(mean))
```
